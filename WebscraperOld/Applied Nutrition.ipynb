```python
import requests
from bs4 import BeautifulSoup
import csv
import time
import pandas as pd
import requests
import numpy as np
from IPython.display import Image, display
import os
```

```python
def is_number(s):
    try:
        float(s)   # works for int, float, "NaN", "inf"
        return True
    except ValueError:
        return False


def AN_find_nutrition(html_text):
    soup = BeautifulSoup(html_text, 'html.parser')
    try:
        name = soup.find("h1", class_ = "heading-bold large").text
    except:
        name = "Name not found"
    try:
        img_tag = soup.find("img", alt="Nutritional Information")
    except:
        return None
    print(name)
    if img_tag:
        img_url = img_tag.get("src") or img_tag.get("data-src")
        display(Image(url=img_url, height=200))
        return img_url, name
    else:
        print("No nutritional information found")

def save_image(img_url, name, save_dir="nutrition_images"):
    """Downloads and saves an image to the specified folder."""
    if not img_url:
        print(f"{name}: No image URL provided.")
        return None

    # Ensure save directory exists
    os.makedirs(save_dir, exist_ok=True)

    # Sanitize filename
    safe_name = "".join(c if c.isalnum() or c in (' ', '-', '_') else "_" for c in name)
    file_path = os.path.join(save_dir, f"{safe_name}.jpg")

    try:
        response = requests.get(img_url)
        response.raise_for_status()
        with open(file_path, "wb") as f:
            f.write(response.content)
        print(f"✅ Saved {name} nutrition image to {file_path}")

        # Optionally display inline (e.g., in Jupyter)
        display(Image(filename=file_path, height=200))
        return file_path

    except Exception as e:
        print(f"❌ Failed to save image for {name}: {e}")
        return None
    
    
```

```python
URL = "https://appliednutrition.uk/collections/best-sellers"
list_page = requests.get(URL)
soup_product_list_page = BeautifulSoup(list_page.text,'html.parser')
product_list = soup_product_list_page.find("ul", id="product-grid")
URL_prefix = "https://appliednutrition.uk/"

# Assuming product_list is the <ul> or container that holds multiple <li class="grid-item"> items
for product_item in product_list.find_all("li", class_="grid-item"):
    # Find the <a> inside <h3 class="card-heading">
    h3_tag = product_item.find("h3", class_="card-heading")
    if not h3_tag:
        continue  # Skip if structure is different

    product_link_tag = h3_tag.find("a")
    if not product_link_tag:
        continue

    product_href = URL_prefix + product_link_tag.get("href")
    print("Product page URL:", product_href)

    # Fetch product page and extract nutrition table
    html_text = requests.get(product_href).text
    if AN_find_nutrition(html_text):
        img_url, name = AN_find_nutrition(html_text)
    save_image(img_url,name,"C:/Users/ryanc/Documents/bs4/Applied Nutrition Images")
    
```
